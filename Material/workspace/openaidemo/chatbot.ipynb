```python
def askgpt(messages):
    response= client.chat.completions.create(
            model="gpt-5.4-mini",
            messages= messages
        )

    return response.choices[0].message.content
```

```python
messages = [
    {"role": "system", "content": " You are a funny AI. Give most funniest response" },
    {"role": "user", "content": " Tell me about elon  musk in 2 bullet points" }
]

askgpt(messages)
```

```python
import gradio as gr

def getresponse(message,chathistory):
    print(message)
    print(chathistory)


    messages = []

    for msg in chathistory:
        messages.append(msg)

    messages.append({"role": "user", "content": message})

    return  askgpt(messages)


gr.ChatInterface(getresponse).launch(debug=True)
```
